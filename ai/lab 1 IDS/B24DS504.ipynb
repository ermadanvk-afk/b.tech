```python
def limited_depth_search(graph, node, goal, limit, path, visited):
    visited.append(node)
    path.append(node)
    if node == goal:
        return list(path)
    if limit <= 0:
        path.pop()
        return None
    for neighbor in graph.get(node, []):
        result = limited_depth_search(graph, neighbor, goal, limit - 1, path, visited)
        if result is not None:
            return result
    path.pop()
    return None
```

```python
def iterative_deepening_search(graph, start, goal, max_depth):
    for depth in range(max_depth + 1):
        visited = []
        path = []

        print("Depth limit:", depth)

        result = limited_depth_search(graph, start, goal, depth, path, visited)

        print("Nodes visited at this limit:", visited)

        if result is not None:
            print("Goal found!")
            print("Path:", result)
            print()
            return result

        print()
    print("Goal not found within max depth =", max_depth)
    return None
```

```python
graph1 = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': ['I'],
    'F': [],
    'G': [],
    'H': [],
    'I': []
}

start1 = 'A'
goal1 = 'I'
iterative_deepening_search(graph1, start1, goal1, max_depth=5)
```

```python
def take_graph_input():
    graph = {}
    n = int(input("Enter number of nodes: "))

    for _ in range(n):
        node = input("Enter node name: ").strip()
        neighbors_line = input("Enter neighbours space separated  (leave blank if none): ").strip()
        if neighbors_line == "":
            neighbors = []
        else:
            neighbors = neighbors_line.split()
        graph[node] = neighbors

    start = input("Enter start node: ").strip()
    goal = input("Enter goal node: ").strip()
    max_depth = int(input("Enter maximum depth to try: "))

    return graph, start, goal, max_depth

user_graph, user_start, user_goal, user_max_depth = take_graph_input()
iterative_deepening_search(user_graph, user_start, user_goal, user_max_depth)
```
