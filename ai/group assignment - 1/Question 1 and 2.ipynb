```python
from collections import deque
import time

def bfs(R,C,matrix,visited1,rs,cs,rg,cg,keep1):
  ans1=[]
  expanded=0
  qu=deque()
  qu.append((rs,cs))
  dr=[-1,0,+1,0]
  dc=[0,+1,0,-1]
  yes=False
  while qu:
    r,c=qu.popleft()
    expanded=expanded+1
    if matrix[r][c]=='G':
      yes=True
      break
    for i in range(4):
      if 0<=r+dr[i]<R and 0<=c+dc[i]<C and matrix[r+dr[i]][c+dc[i]]!='#' and not visited1[r+dr[i]][c+dc[i]]:
        visited1[r+dr[i]][c+dc[i]]=True
        keep1[r+dr[i]][c+dc[i]]=(r,c)
        qu.append((r+dr[i],c+dc[i]))
  if not yes:
    return [],expanded
  r,c=rg,cg
  while r!=rs or c!=cs:
    ans1.insert(0,(r,c))
    r,c=keep1[r][c]
  ans1.insert(0,(rs,cs))
  return ans1,expanded

def get_directions(path):
  ans=[]
  for i in range(1,len(path)):
    r1,c1=path[i-1]
    r2,c2=path[i]
    if r2==r1-1:
      ans.append("UP")
    elif c2==c1+1:
      ans.append("RIGHT")
    elif r2==r1+1:
      ans.append("DOWN")
    else:
      ans.append("LEFT")
  return ans

def dfs(R,C,matrix,visited,rs,cs,rg,cg,ans,expanded):
  expanded[0]=expanded[0]+1
  if rs==rg and cs==cg:
    ans.insert(0,(rs,cs))
    return True
  if rs-1>=0 and matrix[rs-1][cs]!='#' and not visited[rs-1][cs]:
    visited[rs-1][cs]=True
    if dfs(R,C,matrix,visited,rs-1,cs,rg,cg,ans,expanded):
      ans.insert(0,(rs,cs))
      return True
  if cs+1<C and matrix[rs][cs+1]!='#' and not visited[rs][cs+1]:
    visited[rs][cs+1]=True
    if dfs(R,C,matrix,visited,rs,cs+1,rg,cg,ans,expanded):
      ans.insert(0,(rs,cs))
      return True
  if rs+1<R and matrix[rs+1][cs]!='#' and not visited[rs+1][cs]:
    visited[rs+1][cs]=True
    if dfs(R,C,matrix,visited,rs+1,cs,rg,cg,ans,expanded):
      ans.insert(0,(rs,cs))
      return True
  if cs-1>=0 and matrix[rs][cs-1]!='#' and not visited[rs][cs-1]:
    visited[rs][cs-1]=True
    if dfs(R,C,matrix,visited,rs,cs-1,rg,cg,ans,expanded):
      ans.insert(0,(rs,cs))
      return True
  return False

def main():
  R,C=map(int,input().split())
  matrix=[]
  for i in range(R):
    print("Enter row",i+1,":",end=" ")
    row=list(input())
    matrix.append(row)
  rs=0
  cs=0
  rg=0
  cg=0
  for i in range(R):
    for j in range(C):
      if matrix[i][j]=='S':
        rs=i
        cs=j
      if matrix[i][j]=='G':
        rg=i
        cg=j

  visited=[[False for _ in range(C)] for _ in range(R)]
  visited[rs][cs]=True
  expanded=[0]
  ans=[]
  start=time.perf_counter()
  qw=dfs(R,C,matrix,visited,rs,cs,rg,cg,ans,expanded)
  dfs_time=time.perf_counter()-start

  print("Algorithm: DFS")
  print("Path Found:", "Yes" if qw else "No")
  if qw:
    directions=get_directions(ans)
    print("Path:"," ".join(directions))
    print("Number of Moves =",len(directions))
  else:
    print("Path:")
    print("Number of Moves = 0")
  print("Nodes Expanded =",expanded[0])
  print("Execution Time =",dfs_time)

  visited1=[[False for _ in range(C)] for _ in range(R)]
  visited1[rs][cs]=True
  keep1=[[(-1,-1) for _ in range(C)] for _ in range(R)]

  start=time.perf_counter()
  ans1,bfs_expanded=bfs(R,C,matrix,visited1,rs,cs,rg,cg,keep1)
  bfs_time=time.perf_counter()-start

  print("Algorithm: BFS")
  print("Path Found:", "Yes" if ans1 else "No")
  if ans1:
    directions=get_directions(ans1)
    print("Path:"," ".join(directions))
    print("Number of Moves =",len(directions))
  else:
    print("Path:")
    print("Number of Moves = 0")
  print("Nodes Expanded =",bfs_expanded)
  print("Execution Time =",bfs_time)

if __name__=="__main__":
  main()
```

## Comparison and Report

*   **Path length:** Both BFS and DFS found paths for the sample input.
*   **Number of states explored:** DFS expanded fewer nodes 11 compared to BFS 19 for the sample input.DFS might explore fewer nodes if it gets lucky and finds the goal deep in a branch, but BFS explores level by level.
*   **Execution time:** DFS was faster in the sample run.
*   **Optimality of the solution:** BFS is optimal for this problem and is guaranteed to find the shortest path. DFS is not optimal and can find a longer path depending on the search space and traversal order.

```python
import heapq
import time

def gbfs(graph,heurastics,start,goal):
  hq=[]
  visited=set()
  parent={start:None}
  expanded=0
  heapq.heappush(hq,(heurastics[start],start))
  while hq:
    a,b=heapq.heappop(hq)
    if b not in visited:
      visited.add(b)
      expanded+=1
      if b==goal:
        path=[]
        cur=goal
        while cur is not None:
          path.append(cur)
          cur=parent[cur]
        path.reverse()
        return path,expanded
      for i,j in graph[b]:
        if i not in visited:
          if i not in parent:
            parent[i]=b
          heapq.heappush(hq,(heurastics[i],i))
  return [],expanded

def a_star(graph,heurastics,start,goal):
  hq=[]
  visited=set()
  retu=[]
  # current=start
  expanded=0
  heapq.heappush(hq,(heurastics[start],start,0))
  while hq:
    a,b,c=heapq.heappop(hq)
    if b not in visited:
      expanded=expanded+1
      visited.add(b)
      retu.append(b)
    if b==goal:
      return retu,c,expanded
    for i,j in graph[b]:
      if i not in visited:
        heapq.heappush(hq,(c+j+heurastics[i],i,c+j))
  return [],0,0

def main():
  n,m=map(int,input().split())
  edges=[]
  for i in range(m):
    print("Enter edge",i+1,":",end=" ")
    u,v,cost=input().split()
    edges.append((u,v,int(cost)))
  print("Start and end goal node is :")
  start,goal=input().split()
  heurastics={}
  for i in range(n):
    print("Heurastics are :")
    node,h=input().split()
    heurastics[node]=int(h)
  graph={}
  for u,v,cost in edges:
    if u not in graph:
      graph[u]=[]
    if v not in graph:
      graph[v]=[]
    graph[u].append((v,cost))
    graph[v].append((u,cost))

  t=time.perf_counter()
  yes,expanded=gbfs(graph,heurastics,start,goal)
  gbfs_time=time.perf_counter()-t
  print("Algorithm: GBFS")
  print("Path Found:","Yes" if yes else "No")
  if yes:
    print("Path:"," -> ".join(yes))
    cost=0
    for i in range(len(yes)-1):
      for v,c in graph[yes[i]]:
        if v==yes[i+1]:
          cost+=c
          break
    print("Total Path Cost =",cost)
  print("Nodes Expanded =",expanded)
  print("Execution Time =",gbfs_time)

  t=time.perf_counter()
  no,cost,expanded=a_star(graph,heurastics,start,goal)
  a_star_time=time.perf_counter()-t
  print("Algorithm: A*")
  print("Path Found:","Yes" if no else "No")
  if no:
    print("Path:"," -> ".join(no))
    print("Total Path Cost =",cost)
  print("Nodes Expanded =",expanded)
  print("Execution Time =",a_star_time)

if __name__=="__main__":
  main()
```

## Comparison and Report

*   **Path cost:** Both Greedy Best-First Search and A* found paths with a total cost of 10 for the sample input.
*   **Number of states explored:** GBFS expanded fewer nodes 4 compared to A* 5 for the sample input.
*   **Execution time:** A* executed faster in this specific sample run, although GBFS expanded fewer nodes. Execution times can vary based on implementation details and data structures.
*   **Optimality:** A* is optimal (heuristic is admissible) and is guaranteed to find the path with the lowest total cost. GBFS is not optimal as it only considers the heuristic distance to the goal and ignores the cost accumulated so far, which can lead it into dead ends or longer paths.
*   **Effectiveness of the heuristic used:** The heuristic is very effective. It successfully guides both algorithms towards the goal node F, significantly reducing the number of states that need to be explored compared to uninformed searches.
