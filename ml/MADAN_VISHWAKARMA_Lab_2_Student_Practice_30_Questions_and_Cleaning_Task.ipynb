#  Missing Passenger Information

## Story

You are working as a Data Analyst for a maritime research team. The team has received the Titanic passenger dataset and wants to use it for further analysis.

Before any analysis can be trusted, your supervisor asks you to check whether the passenger records are complete.

During an initial inspection, you suspect that some passengers have missing information, particularly regarding their **age, embarkation point, and deck**.

Your task is to investigate the problem and clean the dataset appropriately.

## Your Task

Using the Titanic dataset:

1. Inspect the dataset and identify all columns containing missing values.

2. Determine:
   - The number of missing values in each affected column.
   - The percentage of missing values in each affected column.

3. Create an appropriate visualization to understand the pattern of missing values.

4. Investigate the missing values in the `age`, `embarked`, and `deck` columns.

5. Decide how each missing-value problem should be handled.

6. Apply an appropriate cleaning method to the dataset.

7. Verify whether the missing values have been handled correctly.

8. Explain the reason behind your choice of cleaning method for each column.

## Visualization Requirement

Create at least **one visualization** that clearly shows the distribution or pattern of missing values.

You may use Matplotlib or Seaborn.

## Important

- Do not blindly remove missing records.
- Do not use a pre-written solution.
- Your cleaning decision must be supported by your analysis.
- Explain why your chosen method is appropriate.
- Do not train a machine-learning model.

## Final Question

After completing the cleaning process, answer:

**Is the dataset now suitable for the next stage of data preprocessing and model building? Explain why.**

````python
# inspecting missing values
missing = df.isna().sum()
print("missing values:")
print(missing[missing > 0])

percentage = (missing / len(df)) * 100
print("\npercentage missing:")
print(percentage[percentage > 0])

# plotting missing values
plt.figure(figsize=(10,6))
sns.heatmap(df.isna(), cbar=False, cmap='viridis')
plt.title('missing value pattern')
plt.show()

# fixing the missing values
# age is numeric so replacing with median
df['age'] = df['age'].fillna(df['age'].median())

# embarked is categorical so replacing with mode
mode_embarked = df['embarked'].mode()[0]
df['embarked'] = df['embarked'].fillna(mode_embarked)
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])

# deck has too many missing values so dropping it
df = df.drop(columns=['deck'])

print("\nafter cleaning:\n", df.isna().sum())
````

i used median to fill the missing age values because age is numeric and median is not affected by outliers. for embarked and embark_town, used mode because they are categorical values. for the deck column, dropped the entire column because it had large number of missing values which means it is mostly empty and not useful. 

yes the dataset is now suitable for preprocessing and model building because we handled all the missing values and there are no empty cells left that could cause errors in the machine learning models.

#  The Unusually Expensive Tickets

## Story

You are working as a Data Analyst for a historical maritime research team.

After completing the initial inspection of the Titanic passenger dataset, your supervisor asks you to investigate the **ticket fare** information.

The research team has noticed that most passengers appear to have paid relatively modest fares, while a small number of passengers seem to have paid extremely high fares.

Your supervisor wants to know whether these values are:

- Data-entry errors,
- Unusual but valid observations,
- Or genuine extreme ticket fares.

Before removing anything, you must investigate the data carefully.

## Your Task

Using the `fare` column of the Titanic dataset:

1. Examine the distribution of ticket fares.

2. Calculate:
   - Q1
   - Q2 (Median)
   - Q3
   - IQR

3. Using the IQR method, calculate:

   - Lower Bound
   - Upper Bound

4. Identify the passenger records whose fares fall outside the calculated bounds.

5. Create a **box plot** of the `fare` column to visually identify potential outliers.

6. Inspect the records identified as potential outliers.

7. Decide whether the identified observations should be:
   - Removed,

   - Or investigated further.

8. Justify your decision based on your analysis.

## Visualization Requirement

Create:

Use the visualizations to support your conclusion.

## Important

**Do not automatically remove every value identified as an outlier.**

An outlier is not necessarily an error.

You must investigate the records and explain why you decided to retain or remove them.

## Final Question

After completing your investigation, answer:

**Are the extreme fare values genuine observations or data-quality problems? What decision would you make before using this variable for further analysis? Explain your reasoning.**

````python
# examining fare distribution
plt.figure(figsize=(8,5))
sns.histplot(df['fare'], kde=True)
plt.title('fare distribution')
plt.show()

Q1 = df['fare'].quantile(0.25)
Q2 = df['fare'].quantile(0.50)
Q3 = df['fare'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q2(Median): {Q2}, Q3: {Q3}")
print(f"IQR: {IQR}")
print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")

# identifying outliers
outliers = df[(df['fare'] < lower_bound) | (df['fare'] > upper_bound)]
print(f"\nnumber of outliers: {len(outliers)}")

# boxplot for outliers
plt.figure(figsize=(8,5))
sns.boxplot(x=df['fare'])
plt.title('boxplot of fare')
plt.show()
````

the extreme fare values are genuine observations and not data quality problems. this is because the titanic had different passenger classes and first class tickets were actually very expensive. so these are not data entry errors but real ticket prices.

i would decide to keep these observations for further analysis. if we remove them we will lose important information about the rich passengers and it will limit our analysis for predicting survival based on ticket fare.
